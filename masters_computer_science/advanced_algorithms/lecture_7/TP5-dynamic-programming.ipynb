### Fibonnacci sequence

**Implement a naive recursive algorithm to compute the n-th term of the Fibonnacci sequence.**

The unit test must says that your function seems correct.

```python
%%time
def fibonnaci_naive(n):
    """
    Return the n-th number of the Fibonacci sequence
    n is an integer

    Example:
    >>> fibonnaci_naive(5)
    8
    """
    if n <= 1:
        return 1
    return fibonnaci_naive(n-1) + fibonnaci_naive(n-2)


test_fibonnaci(fibonnaci_naive);
```

**Implement an optimized dynamic programming algorithm to compute the n-th term of the Fibonnacci sequence.**

The unit test must says that your function seems correct.

```python
%%time
def fibonnaci_optimized(n):
    """
    Return the n-th number of the Fibonacci sequence
    n is an integer

    Example:
    >>> fibonnaci_optimized(5)
    8
    """
    arr = [-1]*(n+1)
    arr[0], arr[1] = 1, 1
    for i in range(n+1):
        if i != 0 and i != 1:
            arr[i] = arr[i-1] + arr[i-2]
    return arr[n]

test_fibonnaci(fibonnaci_optimized);
```

### Number decomposition

**Implement a naive recursive algorithm for the number of number decompositions.**

The unit test must says that your function seems correct.

```python
%%time
def number_decomposition_naive(n, mem=None):
  '''
  Return the number of decomposition of n as a sum of the numbers 1, 3 and 4

  Example:
  >>> number_decomposition_naive(5)
  6
  '''
  if mem == None:
    mem = [-1]*(n+1)
    mem[0] = 1
  
  if n == 0:
    return 1
  elif n < 0:
    return 0
  
  if mem[n] != -1:
    return mem[n]

  mem[n] = number_decomposition_naive(n-1, mem) + number_decomposition_naive(n-3, mem) + number_decomposition_naive(n-4, mem)

  return mem[n]
  

test_number_decomposition(number_decomposition_naive);
```

**Implement a dynamic programming algorithm for the number of number decompositions.**

The unit test must says that your function seems correct.

```python
%%time
def number_decomposition_optimized(n, mem=None):
  '''
  Return the number of decomposition of n as a sum of the numbers 1, 3 and 4

  Example:
  >>> number_decomposition_optimized(5)
  6
  '''
  mem = [0]*(n+1)
  mem[0] = 1
  for i in range(n+1):
    if i != 0:
      if i >= 1:
        mem[i] += mem[i-1]
      if i >= 3:
        mem[i] += mem[i-3]
      if i >= 4:
        mem[i] += mem[i-4]
  
  return mem[n]

test_number_decomposition(number_decomposition_optimized);
```

### Rod cutting problem

**Implement a naive recursive algorithm for computing the solution of the rod cutting problem.**

The unit test must says that your function seems correct.

```python
def rod_cutting_cost_naive(n, prices, mem=None):
  """
  Computes the optimal price
  - n is the length of the rod
  - prices is an array of size >=n+1 where p[i] is the price of a rod of length i (p[0]=0)

  return the optimal price

  Example:
  >>> RodCutting(8, [0, 1, 5, 8, 9, 10, 17, 17, 20, 24, 30])
  22
  """
  if mem == None:
    mem = [-1]*(n+1)
  mem[0] = prices[0]
  mem[1] = prices[1]

  if n <= 0:
    return 0
  
  if mem[n] != -1:
    return mem[n]
  
  for i in range(n):
    mem[n] = max(mem[n], prices[n - i] + rod_cutting_cost_naive(i, prices, mem))
  
  return mem[n]
  

test_rod_cutting(rod_cutting_cost_naive, False);
```

**Implement a dynamic programming algorithm for computing the solution of the rod cutting problem.**

The unit test must says that your function seems correct.

```python
def rod_cutting(n, prices):
  """
  Computes the optimal price and the cutting plan
  - n is the length of the rod
  - prices is an array of size >=n+1 where p[i] is the price of a rod of length i (p[0]=0)

  return the optimal price and the cutting plan

  Example:
  >>> RodCutting(8, [0, 1, 5, 8, 9, 10, 17, 17, 20, 24, 30])
  (22, [2, 6])
  """
  mem = [0][0] * (n+1)
  mem = [[0 for _ in range(n+1)] for _ in range(n+1)]
  for i in range(1, n+1):
    for j in range(1, n+1):
      if i < j:
        mem[i][j] = mem[i][i]
      else:
        mem[i][j] = max(mem[i][j-1], prices[j] + mem[i - j][j])

  
  res = mem[n][n]
  cuts = []
  length = n
  while length > 0:
    for cut_length in range(1, length+1):
      if mem[length][length] == prices[cut_length] + mem[length - cut_length][cut_length]:
        cuts.append(cut_length)
        length -= cut_length
        break
  cuts.sort()
  return res, cuts

test_rod_cutting(rod_cutting, True);
```

### Professional robber problem

**Implement a naive recursive algorithm for computing the optimal gain of the professional robber problem.**

The unit test must says that your function seems correct.

```python
%%time
def professional_robber_gain_naive(a, mem=None, n=None):
  """
  Computes the maximal amount of money to steal 
  - a is the array storing the amount of money in each house
  return the maximal amount of money to steal

  Example:
  >>> professional_robber([2, 5, 8, 1, 2, 23, 10, 2, 14, 12])
  47
  """
  if mem == None and n == None:
    n = len(a)-1
    mem = [0]*(n+1)
  
  if n < 0:
    return 0
  elif n == 0:
    return a[0]

  if mem[n] != 0:
    return mem[n]
  
  mem[n] = max(a[n]+professional_robber_gain_naive(a, mem, n-2), professional_robber_gain_naive(a, mem, n-1))

  return mem[n]

test_professional_robber(professional_robber_gain_naive, False);
```

**Implement a dynamic programming algorithm for computing the solution of the professional robber problem.**

The unit test must says that your function seems correct.

```python
%%time
def professional_robber(a):
  """
  Computes the maximal amount of money to steal and the robbing plan (list of robbed houses)
  - a is the array storing the amount of money in each house
  return the maximal amount of money to steal and the robbing plan

  Example:
  >>> professional_robber([2, 5, 8, 1, 2, 23, 10, 2, 14, 12])
  (47, [8, 5, 2, 0])
  """
  n = len(a)
  mem = [0]*(n+1)
  mem[0] = a[0]
  for i in range(1, n):
    mem[i] = max(a[i] + (mem[i-2] if i-2 >= 0 else 0), mem[i-1])
  
  res = mem[n-1]
  plan = []
  i = n-1
  while i >= 0:
    if i == 0:
      plan.append(0)
      break
    if mem[i] == mem[i-1]:
      i -= 1
    else:
      plan.append(i)
      i -= 2
  return res, plan

test_professional_robber(professional_robber, True);
```

### Longest increasing subsequence

**Implement a dynamic programming algorithm for computing the solution of the longest increasing subsequence problem.**

The unit test must says that your function seems correct.

```python
def longest_increasing_subsequence(a, ans=None, n=0):
  '''
  Given a list of integers, return the length and the longest increasing subsequence
  a is the list of integers

  Example:
  >>> longest_increasing_subsequence([1, 4, 3, 1, 2, 4, 3, 2, 8, 5, 2, 7, 5, 10, 8, 9])
  (7, [1, 3, 4, 5, 7, 8, 9])
  '''
  # recursive helper function
  def lis_helper(a, n, mem, prev_index):
    if n == len(a):
      return 0, []
    
    excl_len, excl_seq = lis_helper(a, n+1, mem, prev_index)
    incl_len, incl_seq = 0, []
    if prev_index == -1 or a[n] > a[prev_index]:
      incl_len, incl_seq = lis_helper(a, n+1, mem, n)
      incl_len += 1
      incl_seq = [a[n]] + incl_seq
    if incl_len > excl_len:
      return incl_len, incl_seq
    else:
      return excl_len, excl_seq
  length, sequence = lis_helper(a, 0, {}, -1)
  return length, sequence
  
    

test_longest_increasing_subsequence(longest_increasing_subsequence, True);
```

### Longest common subsequence

**Implement a naive algorithm for computing the solution of the longest common subsequence problem.**

The unit test must says that your function seems correct.

```python
%%time
def LCS_naive(A, B):
  """
  Longest common subsequence between the two strings A and B
  A and B are two strings
  return the longest common subsequence

  Example:
  >>> LCS_naive("azerty", "operazebym")
  'azey'
  """
  if len(A) == 0 or len(B) == 0:
    return ""
  
  if A[-1] == B[-1]:
    return LCS_naive(A[:-1], B[:-1]) + A[-1]
  else:
    left = LCS_naive(A, B[:-1])
    right = LCS_naive(A[:-1], B)
    if len(left) > len(right):
      return left
    else:
      return right

test_LCS(LCS_naive);
```

**Implement a dynamic programming algorithm for computing the solution of the longest common subsequence problem.**

The unit test must says that your function seems correct.

```python
def LCS_dynamic(A, B):
  """
  Longest common subsequence between the two strings A and B
  """
  mem = [["" for _ in range(len(B)+1)] for _ in range(len(A)+1)]
  for i in range(1, len(A)+1):
    for j in range(1, len(B)+1):
      if A[i-1] == B[j-1]:
        mem[i][j] = mem[i-1][j-1] + A[i-1]
      else:
        if len(mem[i-1][j]) > len(mem[i][j-1]):
          mem[i][j] = mem[i-1][j]
        else:
          mem[i][j] = mem[i][j-1]
  return mem[-1][-1]

test_LCS(LCS_dynamic);
```

### Edit distance

**Implement a dynamic programming algorithm for computing the edit distance between two strings.**

The unit test must says that your function seems correct.

```python
def edit_distance(A, B):
  """
  Edit distance between the two strings A and B
  Returns the minimum number of operations to transform A into B

  Example:
  >>> edit_distance("Wiston Cheurchil", "Winston Churchill")
  3
  """
  import time
  start = time.monotonic_ns()
  i, j = len(A), len(B)
  mem = [[-1 for _ in range(j+1)] for _ in range(i+1)]

  def ed_helper(A, B, a, b):
    if a < 0:
      return b + 1
    if b < 0:
      return a + 1
    
    if mem[a][b] != -1:
      return mem[a][b]
    
    if A[a] == B[b]:
      mem[a][b] = ed_helper(A, B, a-1, b-1)
    else:
      mem[a][b] = 1 + min(ed_helper(A, B, a, b-1), ed_helper(A, B, a-1, b), ed_helper(A, B, a-1, b-1))
    return mem[a][b]
  
  ans = ed_helper(A, B, i-1, j-1)
  elapsed = time.monotonic_ns() - start
  print(elapsed)
  return ans

test_edit_distance(edit_distance);
```

### The millionth-visitor

**Implement a dynamic programming algorithm for computing the optimal solution to the millionth-visitor problem.**

The unit test must says that your function seems correct.

```python
def millionth(m, w, p):
  """
  Compute the optimal set of items:
  - m: your weight
  - w: array of item weights
  - p: array of item prices

  Returns the price of the optimal solution and the set of selected items in this optimal solution

  Example:
  >>> millionth(27, [17, 2, 1, 2, 23, 12, 8, 3, 15, 16], [18, 1, 1, 2, 18, 7, 5, 4, 15, 15])
  (26, [8, 4, 3, 2, 1])
  """
  n = len(w)
  mem = [[0 for _ in range(m+1)] for _ in range(n+1)]
  for i in range(1, n+1):
    for j in range(m+1):
      if w[i-1] <= j:
        mem[i][j] = max(mem[i-1][j], p[i-1] + mem[i-1][j - w[i-1]])
      else:
        mem[i][j] = mem[i-1][j]
  
  res = mem[n][m]
  selected_items = []
  weight = m
  for i in range(n, 0, -1):
    if mem[i][weight] != mem[i-1][weight]:
      selected_items.append(i-1)
      weight -= w[i-1]
  selected_items.reverse()
  return res, selected_items

test_millionth(millionth);
```
