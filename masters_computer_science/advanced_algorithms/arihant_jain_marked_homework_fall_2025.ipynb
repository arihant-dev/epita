>Hint : You can do better by sorting the array first, and then picking the element in the right position

>Hint 2: Alternatively, you can use a heap

Don't use build-in median operators

```python
def find_median(array):
    if array == []:
        return None
    for index in range(len(array)):
        smaller_count = 0
        larger_count = 0
        for other_index in range(len(array)):
            if array[other_index] > array[index]:
                smaller_count += 1
            if array[other_index] < array[index]:
                larger_count += 1
            if array[other_index]==array[index]:
                if other_index < index:
                    smaller_count += 1
                if other_index > index:
                    larger_count += 1
                
        if smaller_count == larger_count:
            return array[index]
        
test_find_median(find_median)
## Complexity: O(n^2): We have two nested loops iterating over the array of size n, so the time complexity is O(n^2).
## Space Complexity: O(1).
```

### 3. Maximum k repetitions

Input : A list of integers array, and a repetition limit k, also integer

Output : The same list where the elements can only be repeated k times. 
    With k = 1, this is equivalent to the set of unique elements in the list

>Hint : You can do better by iterating over the array, and keeping the count of how many times each element appears in a dictionnary (map).
    First time you see an element, the dictionnary contains element -> 1, the second time element -> 2, ...
    Then, add each element the appropriate number of times to the result

>Hint 2 : create a new dictionnary: dictionnary = dict()

>Hint 3 : check if it contains a key: if key in dictionnary
          
>Hint 4 : get the value associated with a key: value = dictionnary.get(key)
          
>Hint 5 : put the value associated to a key: dictionnary[key] = value (erases the previous value)

```python
def maximum_k_repetitions(array,k):
    if array == []:
        return []
    array.sort()
    result = []
    index = 0
    while index < len(array):
        max_repetitions = 1
        # we find how many times array[index] repeats
        for j in range(1,k):
            if j+index < len(array) and array[index]==array[index+j]:
                max_repetitions += 1
        # we add it to the result
        for repetition in range(max_repetitions):
            result.append(array[index])
        # we look for a different value
        next_index=index+max_repetitions
        while next_index < len(array) and array[next_index] == array[index]:
            next_index += 1
        index = next_index
        
    return result 

test_max_k_repetitions(maximum_k_repetitions)
## Complexity: Time Complexity - O(n log n): Sorting the array takes O(n log n) time, and the subsequent traversal of the array takes O(n) time.
## Space Complexity: O(n): The result array can store up to n elements in the worst case.
```

```python
def maximum_k_repetitions_fast(array,k):
    if array == []:
        return []
    dict_count = {}
    for element in array:
        if element in dict_count:
            dict_count[element] += 1
        else:
            dict_count[element] = 1
    result = []
    for element, count in dict_count.items():
        repetitions = min(count, k)
        for _ in range(repetitions):
            result.append(element)
    return result

test_max_k_repetitions(maximum_k_repetitions_fast)
## Complexity: Time Complexity - O(n): Counting elements takes O(n) time, and constructing the result also takes O(n) time.
## Space Complexity: O(n): The dictionary and result array can store up to n elements in the worst case.
```

### 7 valid expressions

Input: An array of characters containing only the characters (, ), {, }, [, ], 


Output: true if it is a valid parentheses sequence. 
A sequence is valid if:
- Every opening bracket has a matching closing bracket.
- Brackets close in the correct order.

>Hint: you can use a stack, in a similar way than in exercise 5 here https://perso.esiee.fr/~perretb/AlgorithmDesign/data_structures/index.html

```python
def is_valid_expression(array):
    if isinstance(array, str):
        array = list(array)
    # Keep looping while any removable pair exists
    while True:
        made_change = False

        # Convert list to string for searching
        s_str = ''.join(array)

        if "()" in s_str:
            idx = s_str.index("()")
            # Remove the two characters at that position
            array.pop(idx)
            array.pop(idx)
            made_change = True
        elif "[]" in s_str:
            idx = s_str.index("[]")
            array.pop(idx)
            array.pop(idx)
            made_change = True
        elif "{}" in s_str:
            idx = s_str.index("{}")
            array.pop(idx)
            array.pop(idx)
            made_change = True

        # If no more pairs were removed, stop
        if not made_change:
            break

    return len(array) == 0

test_is_valid_expression(is_valid_expression)

### Complexity : Time Complexity - O(n^2): In the worst case, we may need to scan the entire string multiple times to remove pairs of brackets, leading to a time complexity of O(n^2).
### Space Complexity: O(n): We may need to store the entire string in a list.
```

```python
def is_valid_expression_fast(array):
    stack = []
    for element in array:
        if element == ')':
            if len(stack) > 0 and stack[len(stack)-1] == '(':
                stack = stack[:len(stack)-1]
            else:
                return False
        elif element == '}':
            if len(stack) > 0 and stack[len(stack)-1] == '{':
                stack = stack[:len(stack)-1]
            else:
                return False
        elif element == ']':
            if len(stack) > 0 and stack[len(stack)-1] == '[':
                stack = stack[:len(stack)-1]
            else:
                return False
        else:
            stack.append(element)
        
    return len(stack) == 0

test_is_valid_expression(is_valid_expression_fast)
### Complexity : Time Complexity - O(n) as we are traversing over the array of elements
### Complexity : Space Complexity - O(n) as we have created a new array
```

```python
def split_strings(string_list):
    result = []
    for string in string_list:
        if string in string_list:
            for character in string:
                if character in string:
                    result.append(character)
    return result

test_split_strings(split_strings)

### Complexity: Time Complexity - O(n * k), where k is the length of the largest string in the array of strings, as we are traversing over the array of strings and then on the string as well
### Complexity: Space Complexity - O(n * k), we would create an array which would have a size(in worst case), n*k as all strings could be equal in size with the largest string 
```

```python
def split_strings_fast(string_list):
    if not string_list:
        return []
    result = []
    for string in string_list:
        # append each character of the string to the result using explicit loop
        i = 0
        while i < len(string):
            result.append(string[i])
            i += 1
    return result

test_split_strings(split_strings_fast)

### Complexity: Time Complexity - O(n) where n is total number of characters
# Space Complexity - O(n) as we store all characters in the result
```

### 10 Return grade list 

Input: 
- a list of pairs containing student names and their algorithmics grades, 
- a list of pairs containing student names and their french grades.

Students are not necessarily in the same order

Output: 
- a list of triplets containing student names, their algorithmics grade, their french grade

> Hint A 1 : You can sort using names

> Hint A 2 : to sort a list of pairs using the pairs first elements in python, you can use list.sort(key=lambda x: x[1])


> Hint B 1 : Alternatively, you can also use dictionnaries

```python

def output_student(algo_grades,french_grades):
    result = []
    for name, grade in algo_grades:
        for name_french, grade_french in french_grades:
            if name_french == name:
                result.append((name, grade, grade_french))
    return result

test_output_student(output_student)
#### Complexity:  n = length of algo_grades, m = length of french_grades
#### Time Complexity - O(n*m)
#### Space Complexity - O(min(n, m)) for the result array
```

```python
def output_student_fast(algo_grades,french_grades):
    result = []
    french_grade_dict = {}
    for name_french, grade_french in french_grades:
        french_grade_dict[name_french] = grade_french
    for name, grade in algo_grades:
        if name in french_grade_dict:
            result.append((name, grade, french_grade_dict[name]))
    return result
    
test_output_student(output_student_fast)
#### Complexity:  Time Complexity - O(n + m) where n = length of algo_grades, m = length of french_grades
#### Space Complexity - O(n+m) for the dictionary and result array
```
